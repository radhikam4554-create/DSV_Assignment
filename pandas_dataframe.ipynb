```python
#1.Create a DataFrame named student_df from a dictionary containing the columns Name, Age, and City for
#four students.
student_df = pd.DataFrame({
    "Name": ["Raj", "Priya", "Aman", "Neha"],
    "Age": [20, 21, 19, 22],
    "City": ["Delhi", "Mumbai", "Agra", "Jaipur"]})
```

```python
#2.Display the complete student_df DataFrame.
print(student_df)
```

```python
#3.Write code to display the column names of student_df.
print(student_df.columns)
```

```python
#4.Retrieve the Name column from student_df.
print(student_df["Name"])
```

```python
#5. Retrieve the City column from student_df using square-bracket notation.
print(student_df["City"])
```

```python
#6. Create a new DataFrame from student_df containing only the columns Name and City.
new_df = student_df[["Name", "City"]]
print(new_df)
```

```python
#7. Create a DataFrame named book_df from a dictionary of equal-length lists containing Book, Author, and
#Price.
book_df = pd.DataFrame({
    "Book": ["Python Basics", "Data Science", "Machine Learning"],
    "Author": ["John", "Rahul", "Aman"],
    "Price": [300, 450, 500]
})
print(book_df)
```

```python
#8. Add a new column named Publisher to book_df and assign the same publisher name to every row.
book_df["Publisher"] = "ABC Publications"
print(book_df)
```

```python
#9. Create a five-row DataFrame and assign the values generated by np.arange(5) to a new column named
#SerialNo.
import numpy as np
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"]
})
df["SerialNo"] = np.arange(5)
print(df)
```

```python
#10. Delete the SerialNo column from the DataFrame created in Question 9.
del df["SerialNo"]
print(df)
```

```python
#11. Create a DataFrame named movie_df containing the columns Movie, Year, and Rating.
movie_df = pd.DataFrame({
    "Movie": ["3 Idiots", "Dangal", "PK"],
    "Year": [2009, 2016, 2014],
    "Rating": [8.4, 8.3, 8.1]
})
print(movie_df)
```

```python
#12. From movie_df, create another DataFrame containing the columns Movie, Rating, and a new requested
#column named Director that is not present in the original DataFrame.
movie_new_df = movie_df[["Movie", "Rating"]].copy()
movie_new_df["Director"] = ["Rajkumar Hirani", "Nitesh Tiwari", "Rajkumar Hirani"]
print(movie_new_df)
```

```python
#13. Create a Series containing two city names at indexes 0 and 3.
city_series = pd.Series(
    ["Delhi", "Mumbai"],
    index=[0, 3]
)
print(city_series)
```

```python
#14. Assign the Series from Question 13 to a new column named Branch in a four-row DataFrame.
student_df["Branch"] = city_series
print(student_df)
```
