Objects, Identity (is) vs Equality (==)


```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
```

```python
x = 100
y = 100

print(x is y)
print(x == y)

p = 1000
q = 1000

print(p is q)
print(p == q)
```

Conditions — if, elif, else, modulo (%)

```python
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(0, 5):
    if i % 3 == 0:
        print(i)
    elif i % 3 == 1:
        print(i + 10)
    else:
        print(i - 10)
```

```python
num = float(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```
