Q29. Complete Pandas Series program

```python
import pandas as pd
from pandas import Series

sales = Series(
    [650, 420, 800, 350],
    index=['Laptop', 'Mouse', 'Printer', 'Keyboard']
)

print("Values:")
print(sales.values)

print("Index:")
print(sales.index)

print("Laptop sales:")
print(sales['Laptop'])

print("Sales above 500:")
print(sales[sales > 500])

print("Is Mouse present?")
print('Mouse' in sales.index)

sales_dict = sales.to_dict()
print("Dictionary:")
print(sales_dict)
```

Q30. Series index, missing values, filtering, and names

```python
import pandas as pd
from pandas import Series

data = {'A': 15, 'B': 35, 'C': 25}

s = Series(data, index=['C', 'A', 'D', 'B'])

s.name = 'Observations'
s.index.name = 'Category'

# (a) Display index order
print("Index order:")
print(s.index)

# (b) Display the Series
print("Series:")
print(s)

# (c) Filter values greater than 20
print("Values greater than 20:")
print(s[s > 20])

# (d) Check non-missing values
print("Non-missing values:")
print(pd.notnull(s))

# (e) Display Series name and index name
print("Series name:", s.name)
print("Index name:", s.index.name)
```
